```python
import matplotlib.pyplot as plt
import numpy as np

# 数据准备
full = [10.999999999999996, 18.799999999999997, 45.89999999999999, 55.99999999999999]
refined = [14.7, 41.4, 34.8, 92.5]
categories = ['Full VS BM25 120', 'Full VS BM25 100', 'Full VS BM25 80', 'Full VS BM25 40']  # 使用更短的类别名称

# 创建更窄的图形（宽度减小）
plt.figure(figsize=(7, 5))  # 宽度从10减小到7

# 设置柱状图位置和宽度
x = np.arange(len(categories))
width = 0.3  # 减小柱形宽度

# 绘制柱状图
bars_a = plt.bar(x - width/2, full, width, color='#1f77b4', label='Full')
bars_b = plt.bar(x + width/2, refined, width, color='#ff7f0e', label='Refined')

# 添加数据标签（使用稍小的字体）
plt.bar_label(bars_a, padding=3, fmt='%.1f', fontsize=9)
plt.bar_label(bars_b, padding=3, fmt='%.1f', fontsize=9)

# 设置图形属性
plt.ylabel('Accumulated Sum of Accuracy', fontsize=10)
plt.title('International Law', fontsize=12, fontweight='bold')  # 缩短标题
plt.xticks(x, categories, fontsize=9)  # 使用更小的字体
plt.legend(fontsize=9)  # 使用更小的图例字体

# 添加网格线
plt.grid(axis='y', linestyle='--', alpha=0.7)

# 调整y轴范围以容纳最高柱形
plt.ylim(0, max(max(full), max(refined)) * 1.15)

plt.tight_layout()  # 确保所有元素都显示在画布内
plt.show()
```
