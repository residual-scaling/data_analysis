```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import ast
```

```python
df = pd.read_csv('too_much_repeat.csv')
correctness = df["exp_correctness_record"].to_numpy()
correctness = np.array([ast.literal_eval(s) for s in correctness])
```

```python
# produce the avg score, total length 10
def get_avg(correctness):
    row_count, col_count = correctness.shape
    avg_correctness = []
    for i in range(col_count):
        cur = correctness[:, i]
        avg_correctness.append(float(sum(cur) / row_count) * 10)
    assert len(avg_correctness) == col_count
    return avg_correctness
```

```python
all_avg = get_avg(correctness)
```

```python
print(all_avg)
```

```python
plt.title("Experience")
plt.xlabel("Epoch(s)")
plt.ylabel("Accuracy")
plt.plot(all_avg, marker='o', linestyle='-.', color='limegreen', label='High School Geography')
# plt.plot(hb_avg, marker='s', linestyle='--', label='High School Biology')
# plt.plot(hg_avg, marker='^', linestyle='-.', label='High School Geography')
# plt.plot(sciq_avg, marker='D', linestyle=':', label='SciQ')
plt.legend()
plt.grid()
plt.show()
```
