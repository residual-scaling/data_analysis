```python
import matplotlib.pyplot as plt
import numpy as np

# 计算分数值（转换为小数）
plain = [29/70, 28/70, 33/70, 30/70, 30/70, 28/70, 29/70, 31/70, 27/70, 34/70, 
         33/70, 32/70, 33/70, 32/70, 30/70, 23/70, 33/70, 31/70, 32/70, 31/70]
experience = [36/70, 34/70, 34/70, 36/70, 32/70, 35/70, 34/70, 33/70, 31/70, 34/70, 
              30/70, 34/70, 35/70, 31/70, 34/70, 36/70, 29/70, 33/70, 33/70, 32/70]

# 转换为百分比（可选，根据需求）
# plain = [p * 100 for p in plain]
# experience = [e * 100 for e in experience]

# 创建紧凑的图形
plt.figure(figsize=(10, 5))  # 宽度10高度5，更适合长数据

# 设置柱状图位置和宽度
n = len(plain)  # 数据点数量
x = np.arange(n)
width = 0.35

# 绘制柱状图
bars_plain = plt.bar(x - width/2, plain, width, color='#4C72B0', label='Plain')
bars_exp = plt.bar(x + width/2, experience, width, color='#55A868', label='Experience')

# 添加数据标签（对于重要点可选）
# 由于数据点太多，默认不添加标签以免拥挤
# 但可以添加平均值线

# 计算并添加平均值线
plain_mean = sum(plain)/len(plain)
exp_mean = sum(experience)/len(experience)

plt.axhline(y=plain_mean, color='#4C72B0', linestyle='--', alpha=0.7, linewidth=1.5)
plt.axhline(y=exp_mean, color='#55A868', linestyle='--', alpha=0.7, linewidth=1.5)

# 添加平均值文本
plt.text(n, plain_mean, f' Plain Avg: {plain_mean:.3f}', 
         color='#4C72B0', va='center', ha='left', fontsize=9)
plt.text(n, exp_mean, f' Exp Avg: {exp_mean:.3f}', 
         color='#55A868', va='center', ha='left', fontsize=9)

# 设置图形属性
plt.ylabel('Score (fraction of 70)', fontsize=10)
plt.title('Comparison of Plain vs Experience Scores', fontsize=12, fontweight='bold')
plt.xticks(x, [f'Test {i+1}' for i in range(n)], rotation=45, ha='right', fontsize=8)
plt.legend(fontsize=9)

# 添加网格线
plt.grid(axis='y', linestyle='--', alpha=0.5)

# 调整y轴范围
plt.ylim(0.2, 0.55)  # 根据数据范围设置

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 计算分数值
plain = [29/70, 28/70, 33/70, 30/70, 30/70, 28/70, 29/70, 31/70, 27/70, 34/70, 
         33/70, 32/70, 33/70, 32/70, 30/70, 23/70, 33/70, 31/70, 32/70, 31/70]
experience = [36/70, 34/70, 34/70, 36/70, 32/70, 35/70, 34/70, 33/70, 31/70, 34/70, 
              30/70, 34/70, 35/70, 31/70, 34/70, 36/70, 29/70, 33/70, 33/70, 32/70]

# 创建紧凑的图形
plt.figure(figsize=(10, 5))

# 设置柱状图位置和宽度
n = len(plain)
x = np.arange(n)
width = 0.35

# 绘制柱状图
bars_plain = plt.bar(x - width/2, plain, width, color='#4C72B0', label='Plain')
bars_exp = plt.bar(x + width/2, experience, width, color='#55A868', label='Use Experience')

# 计算平均值
plain_mean = sum(plain)/len(plain)
exp_mean = sum(experience)/len(experience)

# 添加平均值线
plt.axhline(y=plain_mean, color='#4C72B0', linestyle='--', alpha=0.7, linewidth=1.5)
plt.axhline(y=exp_mean, color='#55A868', linestyle='--', alpha=0.7, linewidth=1.5)

# 添加平均值文本 - 向右移动
# 使用图表右侧的空白区域放置文本
plt.text(n + 1.9, plain_mean, f' Plain Avg: {plain_mean:.3f}', 
         color='#4C72B0', va='center', ha='left', fontsize=9)
plt.text(n + 1.9, exp_mean, f' Exp Avg: {exp_mean:.3f}', 
         color='#55A868', va='center', ha='left', fontsize=9)

# 设置图形属性
plt.ylabel('Accuracy', fontsize=10)
plt.title('Accuracy difference over the use of experience', fontsize=12, fontweight='bold')
plt.xticks(x, [f'Round {i+1}' for i in range(n)], rotation=45, ha='right', fontsize=8)
plt.legend(fontsize=9, loc='upper right')  # 将图例放在左上角避免重叠

# 添加网格线
plt.grid(axis='y', linestyle='--', alpha=0.5)

# 调整坐标轴范围，为右侧文本留出空间
plt.xlim(-1, n + 2)  # 左侧留出1个单位空间，右侧留出2个单位空间
plt.ylim(0.2, 0.55)

plt.tight_layout()
plt.show()
```
